```python
import torch

from rlaopt.atoms import ElasticNet, L1Norm, Polyhedron, SumSquares
from rlaopt.expression import Variable
from rlaopt.splitting import ADMMSplit
```

```python
# Variable dimensions
d_x = 10000
d_y = 20000
d_z = 30000

# Number of samples
m = 1000

x = Variable(d_x)
y = Variable(d_y)
z = Variable(d_z)

A_x = torch.randn(m, d_x)
A_y = torch.randn(m, d_y)
A_z = torch.randn(m, d_z)
b = torch.randn(m)

# Use in regularizers
C_y = torch.randn(2, d_y)
C_z = torch.randn(3, d_z)
b_y = torch.randn(2)
b_z = torch.randn(3)

# Use in polyhedral constraint
A_poly = torch.randn(5, d_x)
b_poly = torch.randn(5)
C_poly = torch.randn(4, d_x)
lower_poly = torch.randn(4)
upper_poly = lower_poly + torch.randn(4).abs()
```

```python
loss = (
    SumSquares(A_x @ x + A_y @ y + A_z @ z - b)
    + L1Norm(C_y @ y - b_y)
    + ElasticNet(C_z @ z - b_z)
    + Polyhedron(x, A=A_poly, b=b_poly, C=C_poly, lower=lower_poly, upper=upper_poly)
)
```

```python
admm_split = ADMMSplit(loss)
```
