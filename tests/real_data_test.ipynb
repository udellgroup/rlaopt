```python
import numpy as np
import torch
from rlaopt.expression import Variable
from rlaopt.atoms import LogisticRegression
from rlaopt.data import Dataset, DataLoader
from rlaopt.linalg import NystromConfig
from rlaopt.solvers import Sapphire, SapphireConfig, GradSolverStoppingCriteria
```

```python
from sklearn.datasets import load_svmlight_file
```

```python
X,y = load_svmlight_file("/home/zfran/sapphire/data/libsvm/real-sim")
```

```python
y[y==-1] =0 
```

```python
X = X.toarray()
```

```python
dataset = Dataset(X,y, device="cuda:0")
```

```python
loader = DataLoader(dataset, batch_size=1024)
```

```python
beta = Variable((X.shape[1]), name="beta", device="cuda:0")
```

```python
model = LogisticRegression(beta, loader, fit_intercept=False)
```

```python
obj = model
```

```python
prec_config = NystromConfig(
    rank_init=10, error_tolerance=1.0, base_damping=1e-2, damping_mode="adaptive"
)
config = SapphireConfig(eta0=0.1, grad_batch_size=1024, subproblem_iters=20)
opt = Sapphire(obj, config)
```

```python
stop_criteria = GradSolverStoppingCriteria(max_iters=1000)
```

```python
result = opt.solve(stopping_criteria=stop_criteria)
```
